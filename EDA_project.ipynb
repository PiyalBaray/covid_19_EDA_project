```python
conn = http.client.HTTPSConnection("covid-19-statistics.p.rapidapi.com")

headers = {
    'x-rapidapi-key': "ab8cd74ce5msha3a4f3decdc126ap1accf1jsnfdaa84a503eb",
    'x-rapidapi-host': "covid-19-statistics.p.rapidapi.com"
}

conn.request("GET", "/provinces?iso=AUS", headers=headers)

res = conn.getresponse()
data = res.read()

print(data.decode("utf-8"))
```

```python


# Since data already has the response bytes, decode and load them directly as JSON
json_payload = json.loads(data.decode('utf-8'))

# Extract the nested 'data' list and structure it as a clean DataFrame
df = pd.DataFrame(json_payload['data'])
df.head()
```
